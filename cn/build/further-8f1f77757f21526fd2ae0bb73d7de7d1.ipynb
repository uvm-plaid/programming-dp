# 延伸阅读

差分隐私已经发展成为一个成熟的交叉学科领域，涵盖理论、系统和实践。本章重点介绍一些奠基性论文、有影响力的系统和活跃的研究方向，供希望更进一步的读者参考！

## 差分隐私的理论基础

- 差分隐私的原始定义和早期机制：
  - Dwork等人（2006）：*Calibrating Noise to Sensitivity in Private Data Analysis* {cite}`dwork2006B`
  - Dwork（2008）：*Differential Privacy: A Survey of Results* {cite}`dwork2008survey`


## 实用系统与实现

- 生产级差分隐私系统：
  - 谷歌的[RAPPOR](https://research.google/pubs/pub42852/)和苹果的[差分隐私部署](https://www.apple.com/privacy/docs/Differential_Privacy_Overview.pdf)

- 工具库与框架：
  - [Opacus](https://opacus.ai/)：用于在PyTorch中实现差分隐私
  - [TensorFlow Privacy](https://github.com/tensorflow/privacy)

## 差分隐私机器学习

- DP-SGD与现代差分隐私机器学习：
  - Abadi等人（2016）：*Deep Learning with Differential Privacy* {cite}`abadi2016deep`
  - Bu等人（2022）：*Deep Learning with Gaussian Differential Privacy* {cite}`bu2022deep`

- 隐私攻击与审计：
  - Shokri等人（2017）：*Membership Inference Attacks* {cite}`shokri2017membership`
  - Carlini等人（2019）：*Secret Sharer: Evaluating and Testing Unintended Memorization* {cite}`carlini2019secret`

## 实验设计与统计

- 用于统计估计的差分隐私：
  - Smith（2011）：*Privacy-Preserving Statistical Estimation with Optimal Convergence Rates* {cite}`smith2011privacyestimation`
  - Karwa和Slavkovic（2016）：*Inference in Contingency Tables under Differential Privacy* {cite}`karwa2016noisydegrees`

- 差分隐私假设检验：
  - Sheffet（2017）：*Differentially Private Ordinary Least Squares* {cite}`sheffet2017dpols`

## 进阶主题

- 本地差分隐私：
  - Kasiviswanathan等人（2011）：*What Can We Learn Privately?* {cite}`kasiviswanathan2011learnprivately`
  - Erlingsson等人（2014）：*RAPPOR: Privacy-Preserving Reporting System* {cite}`rappor`

- 组合性与隐私核算：
  - Mironov（2017）：*Rényi Differential Privacy* {cite}`mironov2017renyi`
  - Kairouz等人（2015）：*The Composition Theorem in Differential Privacy* {cite}`kairouz2015composition`

- 验证与形式化推理：
  - Reed, Jason和Pierce, Benjamin C.（2010）：*Distance makes the types grow stronger: a calculus for differential privacy* {cite}`pierce2010distance`
  - Abuah等人（2022）：*Solo: a lightweight static analysis for differential privacy* {cite}`abuah2022solo`
