## 2. 汇总统计

设计一种算法来生成满足差分隐私的下述统计数据：

- 均值：$\mu = \frac{1}{n} \sum_{i=1}^n x_i$
- 方差：$var = \frac{1}{n} \sum_{i=1}^n (x_i - \mu)^2$
- 标准差：$\sigma = \sqrt{\frac{1}{n} \sum_{i=1}^n (x_i - \mu)^2}$

**思想**：

**均值**

1. 使用稀疏向量技术找到裁剪边界的上界和下界
2. 计算噪声求和值与噪声计数值，再应用后处理性得到均值

**方差**

1. 将方差问询拆分为一个计数问询（并计算得到$\frac{1}{n}$，我们可以从计算均值问询的过程中得到计数问询的结果）与一个求和问询
2. $\sum_{i=1}^n (x_i - \mu)^2$的敏感度是什么？答案是$b^2$；我们可以裁剪并计算$\sum_{i=1}^n (x_i - \mu)^2$，然后根据后处理性与（1）得到的结果相乘

**标准差**

1. 只需计算方差的平方根

涉及到的全部问询：
- 裁剪下界（稀疏向量技术）
- 裁剪上界（稀疏向量技术）
- 噪声求和（均值）
- 噪声计数
- 噪声求和（方差）

## 4. 分层问询

设计一种为美国人口普查信息生成汇总统计结果的算法。算法应该能按下述从低到高的层次输出相应的人口统计结果：

- 人口普查区
- 城市 / 乡镇
- 邮编
- 县
- 州
- 美国

```{tip}

思想1：使用并行组合性，*只*计算最低层次（人口普查区）的人口统计结果。将所有区域的计数结果相加，得到各个城市的人口统计结果，以此类推，得到所有统计结果。优势：隐私预算低。

思想2：计算所有层次的计数结果，分别对每一层使用并行组合性，根据真实数据调整预算分配。对于层次结构中较小的层级，我们可能需要更高的准确性。

思想3：和（2）一样，但应用后处理性，基于更高层的计数结果重新缩放较低层的计数结果，将缩放后的（浮点数）结果截断为整数；将负数设置为0。

```

## 5. 一系列范围问询

设计一种算法来准确回答一系列*范围问询*。这些范围问询都是针对某一个数据表的问询："有多少行数据的值在$a$和$b$之间？"（即特定取值范围的数据行数）。

### 第1部分

这一系列范围问询是已经预先确定的、数量有限的、形式为： $\{(a_1, b_1), \dots, (a_k, b_k)\}$的问询序列。

### 第2部分

这一系列范围问询序列的长度$k$是预先确定的，但是问询以流方式执行，每一个问询必须在执行时就给出回复。

### 第3部分

范围问询序列可能是无限长的。

**思想**：

根据串行组合性，依次执行每一个问询

对于第1部分，我们可以引入$L2$敏感度，从而引入高斯机制。当$k$很大时，高斯噪声的应用效果会更好。

或者，我们可以构造合成数据：

- 为每个问询范围$(i, i+1)$计算一个计数值（这样就可以应用并行组合性了）。这就是所谓的合成数据表示法！我们可以将直方图中落在指定问询区间内的所有分段计数结果相加，从而回答无穷多的范围问询。
- 对于第2部分，使用稀疏向量技术

使用稀疏向量技术：对于问询流中的每个问询，查看真实数据回复结果与合成数据回复结果之间的差值。如果差值较大，则问询一次真实数据，得到（应用并行组合性，得到直方图形式的）回复结果，并更新合成数据。否则，给出直接返回合成数据回复结果。这样一来，只有当需要更新合成数据时，我们*才需要*消耗隐私预算。

## 隐私部署检查清单

部署差分隐私需要执行若干步骤，以确保在保护敏感数据的同时，仍然能够进行有用的数据分析。下面是一份部署差分隐私系统的检查清单：

1. **数据分类**：识别需要保护的敏感数据，并对其进行分类。了解每类数据的敏感程度及其潜在的风险。这一步应该包括确定[隐私单位](#unit-of-privacy)，我们之前已经讨论过这个概念。

2. **数据准备**：预处理数据，移除个人身份信息（Personally Identifiable Information，PII）以及与分析无关的其他敏感属性。

3. **确定隐私预算**：确定部署中可接受的效用水平或隐私风险水平。这将决定为实现差分隐私需要向数据添加多少噪声。

4. **数据分析需求**：清晰地列出数据分析目标，以及在保护隐私的前提下需要支持的具体数据变换和问询。确保所选择的隐私机制和隐私单位能够满足这些需求。

5. **实现差分隐私**：根据分析需求，选择合适的差分隐私机制，如拉普拉斯机制、高斯机制或其他高级技术，并将这些机制集成到数据处理流水线中。如有需要，确定隐私组合的策略。

6. **噪声生成**：根据所选择的差分隐私机制，生成高质量的熵并将其引入数据中。确保随机性水平经过校准，可以实现预期的隐私保证。

7. **测试与验证**：进行全面的测试，验证所部署差分隐私机制的有效性。使用各种问询和场景测试系统，确保在保持数据效用的同时保护隐私。理想情况下，应在公开数据或合成数据上进行测试。

8. **性能评估**：评估差分隐私机制的时间和存储开销所带来的性能损失。监控延迟、吞吐量和计算资源利用率等系统指标，确保效率处于可接受的水平。

9. **文档与合规**：记录部署过程，包括所使用的差分隐私机制、所选择的隐私参数以及其他相关细节。确保符合相关的隐私法律、法规和标准。

10. **额外的安全措施**：实施所有必要的安全措施，防范潜在的攻击或漏洞。这可能包括对传输中和静态存储的数据进行加密、访问控制以及审计机制，也可能包括防范侧信道攻击的保护措施，如隔离的计算环境。

11. **用户教育与培训**：向用户和利益相关者介绍差分隐私的原理及其影响，并让他们认识到在数据分析过程中保护隐私的重要性。

12. **持续监控与维护**：建立一套对已部署系统进行持续监控和维护的流程。定期审查隐私参数，并根据需要更新机制，以适应不断变化的隐私需求和威胁。

遵循这份检查清单，您就可以有效地部署一个差分隐私系统，在保护敏感数据的同时获得有价值的分析结果和见解！
